```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
clients = pd.read_csv("../data/raw/clients.csv")
properties = pd.read_csv("../data/raw/properties.csv")

print("Clients dataset loaded successfully.")
print("Properties dataset loaded successfully.")
```

```python
missing_clients = clients.isnull().sum().to_frame(name="Missing Values")
missing_clients["Percentage"] = (missing_clients["Missing Values"] / len(clients)) * 100
missing_clients
```

```python
missing_properties = properties.isnull().sum().to_frame(name="Missing Values")
missing_properties["Percentage"] = (missing_properties["Missing Values"] / len(properties)) * 100
missing_properties
```

```python
print("Client duplicates:", clients.duplicated().sum())
print("Property duplicates:", properties.duplicated().sum())
```

```python
for col in clients.columns:
    print(f"\\n{col}")
    print(clients[col].nunique())
```

```python
print("Unique client IDs:", clients["client_id"].nunique())
print("Unique property client references:", properties["client_ref"].nunique())

matched = properties["client_ref"].isin(clients["client_id"]).sum()

print("Matching property records:", matched)
print("Matching percentage:", round(matched / len(properties) * 100, 2), "%")
```

## Initial observations

- The client dataset contains demographic and financing attributes.
- The property dataset contains investment transaction information.
- The datasets can potentially be merged using `client_id` and `client_ref`.
- Missing values and duplicate records will be handled during preprocessing.
- Date of birth will be transformed into age.
- Categorical variables will require encoding for clustering.
