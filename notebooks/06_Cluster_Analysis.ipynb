# Buyer segment analysis and business intelligence

This notebook analyzes the buyer segments generated through clustering and produces business intelligence insights for real estate market strategy, customer targeting, financing analysis, and geographic investment profiling.

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

pd.set_option('display.max_columns', None)

buyer_profile = pd.read_csv('../data/processed/buyer_segmented_dataset.csv')

print(buyer_profile.shape)
buyer_profile.head()
```

```python
segment_summary = buyer_profile.groupby('buyer_segment').agg(
    Buyers=('client_id', 'count'),
    Average_Age=('age', 'mean'),
    Average_Satisfaction=('satisfaction_score', 'mean'),
    Average_Properties=('total_properties_owned', 'mean'),
    Average_Investment=('total_investment_value', 'mean'),
    Average_Property_Value=('average_property_value', 'mean'),
    Average_Floor_Area=('average_floor_area', 'mean')
).round(2)

segment_summary
```

```python
segment_counts = (
    buyer_profile['buyer_segment']
    .value_counts()
    .reset_index()
)

segment_counts.columns = ['Buyer Segment', 'Buyers']

fig = px.bar(
    segment_counts,
    x='Buyer Segment',
    y='Buyers',
    color='Buyer Segment',
    text='Buyers',
    title='Distribution of Buyer Segments'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
fig = px.violin(
    buyer_profile,
    x='buyer_segment',
    y='age',
    color='buyer_segment',
    box=True,
    points='outliers',
    title='Age Distribution by Buyer Segment'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
gender_segment = (
    buyer_profile
    .groupby(['buyer_segment', 'gender'])
    .size()
    .reset_index(name='Count')
)

fig = px.bar(
    gender_segment,
    x='buyer_segment',
    y='Count',
    color='gender',
    barmode='group',
    title='Gender Distribution Across Buyer Segments'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
client_segment = (
    buyer_profile
    .groupby(['buyer_segment', 'client_type'])
    .size()
    .reset_index(name='Count')
)

fig = px.bar(
    client_segment,
    x='buyer_segment',
    y='Count',
    color='client_type',
    barmode='stack',
    title='Client Type Composition by Buyer Segment'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
fig = px.violin(
    buyer_profile,
    x='buyer_segment',
    y='total_investment_value',
    color='buyer_segment',
    box=True,
    points='outliers',
    title='Investment Value Distribution by Buyer Segment'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
portfolio_summary = (
    buyer_profile
    .groupby('buyer_segment')['total_properties_owned']
    .mean()
    .reset_index()
)

fig = px.bar(
    portfolio_summary,
    x='buyer_segment',
    y='total_properties_owned',
    color='buyer_segment',
    text='total_properties_owned',
    title='Average Properties Owned by Buyer Segment'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
property_summary = (
    buyer_profile
    .groupby('buyer_segment')['average_property_value']
    .mean()
    .reset_index()
)

fig = px.bar(
    property_summary,
    x='buyer_segment',
    y='average_property_value',
    color='buyer_segment',
    text='average_property_value',
    title='Average Property Value by Buyer Segment'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
loan_segment = (
    buyer_profile
    .groupby(['buyer_segment', 'loan_applied'])
    .size()
    .reset_index(name='Count')
)

fig = px.bar(
    loan_segment,
    x='buyer_segment',
    y='Count',
    color='loan_applied',
    barmode='group',
    title='Loan Dependency Across Buyer Segments'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
loan_rate = (
    buyer_profile
    .assign(loan_yes=buyer_profile['loan_applied'].str.lower().eq('yes').astype(int))
    .groupby('buyer_segment')['loan_yes']
    .mean()
    .reset_index()
)

loan_rate['Loan Rate (%)'] = loan_rate['loan_yes'] * 100

fig = px.bar(
    loan_rate,
    x='buyer_segment',
    y='Loan Rate (%)',
    color='buyer_segment',
    text='Loan Rate (%)',
    title='Loan Application Rate by Buyer Segment'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
purpose_segment = (
    buyer_profile
    .groupby(['buyer_segment', 'acquisition_purpose'])
    .size()
    .reset_index(name='Count')
)

fig = px.bar(
    purpose_segment,
    x='buyer_segment',
    y='Count',
    color='acquisition_purpose',
    barmode='stack',
    title='Acquisition Purpose by Buyer Segment'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
region_segment = (
    buyer_profile
    .groupby(['region', 'buyer_segment'])
    .size()
    .reset_index(name='Count')
)

fig = px.bar(
    region_segment,
    x='region',
    y='Count',
    color='buyer_segment',
    title='Regional Distribution of Buyer Segments'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
country_segment = (
    buyer_profile
    .groupby(['country', 'buyer_segment'])
    .size()
    .reset_index(name='Count')
)

top_countries = (
    country_segment
    .groupby('country')['Count']
    .sum()
    .nlargest(10)
    .index
)

filtered = country_segment[country_segment['country'].isin(top_countries)]

fig = px.bar(
    filtered,
    x='country',
    y='Count',
    color='buyer_segment',
    title='Top Countries by Buyer Segment'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
fig = px.box(
    buyer_profile,
    x='buyer_segment',
    y='satisfaction_score',
    color='buyer_segment',
    title='Customer Satisfaction by Buyer Segment'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
referral_segment = (
    buyer_profile
    .groupby(['buyer_segment', 'referral_channel'])
    .size()
    .reset_index(name='Count')
)

fig = px.bar(
    referral_segment,
    x='buyer_segment',
    y='Count',
    color='referral_channel',
    title='Referral Channel Effectiveness by Buyer Segment'
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
# Create loan_flag if it does not exist
buyer_profile['loan_flag'] = (
    buyer_profile['loan_applied']
    .astype(str)
    .str.lower()
    .eq('yes')
    .astype(int)
)

# Verify
buyer_profile[['loan_applied', 'loan_flag']].head()
```

```python
executive_summary = buyer_profile.groupby('buyer_segment').agg(
    Buyers=('client_id', 'count'),
    Average_Investment_Value=('total_investment_value', 'mean'),
    Average_Properties_Owned=('total_properties_owned', 'mean'),
    Loan_Application_Rate=('loan_flag', 'mean'),
    Average_Satisfaction=('satisfaction_score', 'mean'),
    Average_Age=('age', 'mean')
).round(2)

# Convert loan rate to percentage
executive_summary['Loan_Application_Rate'] = (
    executive_summary['Loan_Application_Rate'] * 100
).round(1)

executive_summary
```
